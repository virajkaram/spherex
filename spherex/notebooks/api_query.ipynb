```python
import requests
```

```python
API_URL = "http://irtda.mit.edu:88"
USERNAME = "x   x"
PASSWORD = "- - -"

auth = (USERNAME, PASSWORD)
```

```python
# --- 1️⃣ Submit a new job ---
payload = {
    "targets": [
        {"ra":  11.298702, "dec": 41.419959, "targetname": "WNTR24okvms"},
        {"ra": 9.703196,  "dec": 40.768867, "targetname": "WNTR23bdbvo"}
    ]
}

r = requests.post(f"{API_URL}/submit", json=payload, auth=auth)
r.raise_for_status()
job = r.json()
print("Submitted job:", job)
job_id = job["id"]
```

```python
# Poll job status until complete ---
r = requests.get(f"{API_URL}/job/{job_id}", auth=auth)
r.raise_for_status()
info = r.json()
print("Job status:", info["status"])
```

```python
# Retrieve results ---
r = requests.get(f"{API_URL}/retrieve/{job_id}", auth=auth)
if r.status_code == 200:
    with open(f"job_{job_id}_output.zip", "wb") as f:
        f.write(r.content)
    print(f"Downloaded job output: job_{job_id}_output.zip")
else:
    print("Retrieve failed:", r.text)
```

```python
# Unzip the directory
import zipfile
import os
with zipfile.ZipFile(f"job_{job_id}_output.zip", 'r') as zip_ref:
    output_dir = f"job_{job_id}_output"
    os.makedirs(output_dir, exist_ok=True)
    zip_ref.extractall(output_dir)
    print(f"Extracted to directory: {output_dir}")
```

```python
# Read and plot the spectrum
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

spectrum_file = os.path.join(output_dir, "spectrum_WNTR24okvms_ra11.29870_dec41.41996.csv")
spectrum_df = pd.read_csv(spectrum_file)
```

```python
plt.figure()

plt.errorbar(spectrum_df['wavelength_um'], spectrum_df['flux_bkgsub_jy']*1e6,
             yerr=spectrum_df['flux_bkgsub_err_jy']*1e6, fmt='o', label='SPHEREx Spectrum')

flagged = spectrum_df[spectrum_df['flagged'] == True]
plt.scatter(flagged['wavelength_um'], flagged['flux_bkgsub_jy']*1e6, color='red', marker='x', s=50, label='Flagged Points')

plt.legend()
plt.xlabel('Wavelength (um)')
plt.ylabel('Flux density (uJy)')
plt.ylim(-500, 4000)
plt.xscale('log')
```
